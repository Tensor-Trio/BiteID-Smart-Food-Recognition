Hyperparameter experiments, learning curves, and model comparison logs.

```python
import json
import matplotlib.pyplot as plt
import pandas as pd

# Load training histories
with open('../models/baseline_history.json') as f:
    baseline_hist = json.load(f)

with open('../models/transfer_history.json') as f:
    transfer_hist = json.load(f)

print('Baseline keys:', baseline_hist.keys())
print('Transfer keys:', transfer_hist.keys())
```

```python
# Plot learning curves
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Baseline
axes[0,0].plot(baseline_hist['loss'], label='Train Loss')
axes[0,0].plot(baseline_hist['val_loss'], label='Val Loss')
axes[0,0].set_title('Baseline CNN - Loss')
axes[0,0].legend()
axes[0,0].grid(True)

axes[0,1].plot(baseline_hist['accuracy'], label='Train Acc')
axes[0,1].plot(baseline_hist['val_accuracy'], label='Val Acc')
axes[0,1].set_title('Baseline CNN - Accuracy')
axes[0,1].legend()
axes[0,1].grid(True)

# Transfer
axes[1,0].plot(transfer_hist['loss'], label='Train Loss')
axes[1,0].plot(transfer_hist['val_loss'], label='Val Loss')
axes[1,0].set_title('Transfer Learning - Loss')
axes[1,0].legend()
axes[1,0].grid(True)

axes[1,1].plot(transfer_hist['accuracy'], label='Train Acc')
axes[1,1].plot(transfer_hist['val_accuracy'], label='Val Acc')
axes[1,1].set_title('Transfer Learning - Accuracy')
axes[1,1].legend()
axes[1,1].grid(True)

plt.tight_layout()
plt.show()
```
